#### **Introduction**

#### **AI in Transport**

Early applications of AI in transport focused on rule-based expert systems for traffic signal control and route optimisation. However, the rise of machine learning and deep learning has dramatically expanded the scope of what is possible. #### **Computer Vision for Vehicle Detection**

Computer Vision is arguably the most impactful AI subfield in transportation. The ability to extract meaningful information from images and video streams has enabled a wide range of practical systems.

Early approaches to vehicle detection relied on handcrafted features, such as the Histogram of Oriented Gradients (HOG), combined with Support Vector Machines (SVMs). While effective under constrained conditions, these methods struggle with variations in lighting, occlusion, and vehicle diversity (Papageorgiou & Poggio, 2000).

The breakthrough came with Convolutional Neural Networks (CNNs). LeCun et al. (1998) demonstrated that CNNs could learn spatial feature hierarchies directly from raw image data, eliminating the need for manual feature engineering. This capability proved transformative for vehicle detection.

A major milestone was the development of Region-based CNN (R-CNN) by Girshick et al. (2014), which combined region proposals with CNN-based feature extraction to achieve state-of-the-art object detection. Subsequent improvements – Fast R-CNN and Faster R-CNN – made these systems practical for real-time applications.

The introduction of the You Only Look Once (YOLO) family of models (Redmon et al., 2016) was a further turning point. YOLO frames object detection as a single regression problem, processing an entire image in a single forward pass through the network. This allows detection at real-time traffic camera speeds. Subsequent YOLO versions (v3 through v8 and beyond) have continued to improve accuracy and speed, with YOLOv5 and YOLOv8 now widely used in industry.

#### **Notable Research and Developments**

Several notable studies demonstrated the effectiveness of CNN-based vehicle detection in transportation:

* Traffic monitoring
  
  Lou et al. (2019) applied a Faster R-CNN model to a traffic surveillance dataset, achieving over 90% accuracy in detecting and classifying vehicles by type (car, truck, bus, motorcycle). The system was deployed in a smart city pilot in Shenzhen, China, enabling dynamic signal control based on real-time traffic counts.

* Autonomous driving

  The KITTI Vision Benchmark Suite (Geiger et al., 2012) has become a standard evaluation framework for 3D object detection in autonomous driving. Models trained on KITTI must detect vehicles, pedestrians, and cyclists from LiDAR and camera data simultaneously – a task where deep learning has consistently outperformed classical methods.

* Highway incident detection

  Kamijo et al. (2000) used spatio-temporal analysis of video feeds to detect traffic incidents on motorways. More recently, transformer-based vision models have been applied to the same problem with significantly higher detection rates and fewer false positives.

* Aerial and drone-based detection

  VisDrone (Zhu et al., 2018) is a large-scale benchmark for object detection from drone footage, covering diverse urban scenarios. CNNs fine-tuned on this dataset have demonstrated the ability to detect small vehicles at high altitudes, with applications in urban planning and emergency response.

#### **Challenges and Limitations**

Despite significant progress, AI-based vehicle detection in transportation still faces several challenges:

* **Occlusion**: Vehicles partially hidden by other objects or infrastructure remain difficult to detect reliably.

* **Small object detection**: Vehicles at a distance or viewed from an aerial perspective appear very small in the image, reducing detection accuracy.

* **Domain shift**: Models trained on datasets from one city or climate often perform poorly when deployed in a different environment due to differences in road layout, vehicle types and lighting.

* **Real-time constraints**: Edge deployment on embedded hardware (e.g., traffic cameras) requires highly optimised, lightweight models.

* **Annotation cost**: Training deep learning models requires large amounts of labelled data, which is expensive and time-consuming to produce.

Researchers have addressed some of these challenges through transfer learning, data augmentation, and model compression techniques. The use of pre-trained models such as ResNet and EfficientDet has significantly reduced the data and computational requirements for new deployments.

#### **Convolution Neural Network (CNN)**

CNNs are a class of deep learning model specifically designed for processing grid-structure data such as images. They use learned convolutional filters to extract spatial features - edges, shaed, textures - at multiple levels of abstraction. YOLOv5 is a CNN-based architecture that performs object detection in a single forward pass, treating detection as a regression problem over a grid of cells (Redmin et al., 2016).

* **Input data required:**

  Raw RGB images (e.g., JPEG or PNG), typically resized to 416x416 or 640x640 pixels. For training, annotated images with bounding box coordinates and class labels in YOLO format are required. Pre-trained weights trained on the COCO dataset can be used for transfer learning, significantly reducing data requirements.

* **Expected output:**

  Bounding boxes drawn around each detected vehicle, class label (car, truck, bus, motorcycle), and a confidence score (0-1) for each detection. Multiple detections are produced per image in a single pass.

* **Application to vehicle detection:**

  YOLO can process a full traffic camera and simultaneously detect all vehicles present. A model pre-trained on COCO (which already includes car, truck, bus, and motorcycle classes) can be fined-tuned or used directly. YOLOv5 acheives detection speeds of 30-100+ frames per second on a modern GPU, making it viable for live traffic monitoring.

* **Strengths:**

  Extremely high accuracy with state-of-the-art performance on standard benchmarks (COCO mAP > 50%). Supports real-time detection. Detects multiple objects of different classes simultaneously. Transfer learning reduces data and training time significantly. Handles scale variation, partial occlusion, and cluttered backgrounds well.

* **Weaknesses:**

  Requires significant computational resources (GPU recommended for training). Black-box nature makes predictions difficult to interpret. Requires large annotated datasets when training from scratch. Hyperparameter tuning can be complex for beginners. Risk of overfitting on small datasets without regularisation.

#### **Support Vector Machine (SVM) with HOG Features**

A Support Vector Machine is a supervised machine learning classifier that finds the optimal hyperplane to separate classes in a high-dimensional feature space. For image-based tasks, SVMs are typically combined with hand-crafted feature descriptors. The most common for vehicle detection is the Histogram of Oriented Gradients (HOG), which encodes the distribution of edge directions in local image regions. This was the dominant approach to pedestrian and vehicle detection before deep learning became prevalent (Dalal and Triggs, 2005).

* **Input data required:**

  Fixed-size image patches (e.g., 64×128 pixels) extracted via a sliding window across the image. HOG feature vectors computed from each patch (typically 1,764-dimensional for a 64×128 patch with standard parameters). Moderate dataset size required: hundreds to low thousands of samples per class.

* **Expected output:**

  Binary or multi-class classification per image patch (vehicle / not vehicle). Bounding boxes are obtained by combining the sliding window approach with non-maximum suppression post-processing.

* **Application to vehicle detection:**

  An SVM classifier trained on HOG features scans a traffic camera image using a sliding window at multiple scales, classifying each window as containing a vehicle or not. Overlapping positive detections are merged using non-maximum suppression. This pipeline was widely used in early vehicle detection systems and remains in use in resource-constrained embedded deployments where a GPU is unavailable.

* **Strengths:**

  Well-established, theoretically grounded method with strong mathematical guarantees. Works effectively with smaller datasets. Computationally lightweight and can run on CPU without specialised hardware. More interpretable than deep neural networks. Fast training time compared to CNNs. Robust to overfitting with correct regularisation.

* **Weaknesses:**

  HOG features are hand-crafted and do not adapt to the data, making them inferior to learned CNN features. The sliding window approach is computationally expensive at inference time. Struggles with large variations in viewpoint, lighting, and vehicle appearance. Does not naturally handle multiple object detection in one pass. Performance degrades significantly compared to CNNs on complex, real-world scenes.

#### **K-Nearest Neighbours (KNN) with Image Features**

K-Nearest Neighbours is a simple, non-parametric supervised learning algorithm. During inference, a new data point is classified by finding the K most similar training examples and taking a majority vote of their labels (Cover and Hart, 1967). For image classification, each image is represented as a feature vector — either raw pixel values, colour histograms, or descriptors such as HOG or ORB. KNN requires no explicit training phase; the entire dataset is stored and used directly at inference time.

* **Input data required:**

  Fixed-size images (e.g., 64×64 pixels) flattened or converted to a feature vector. A labelled training set stored in memory for comparison at inference time. Dataset size is flexible, but accuracy is highly dependent on the quality and coverage of training examples.

* **Expected output:**

  Class label for each input image patch (car, truck, bus, background). Optionally a probability estimate based on the proportion of nearest neighbours per class. Works at patch level; full-image detection requires sliding window integration.

* **Application to vehicle detection:**

  KNN can be applied by extracting HOG or colour histogram features from fixed-size image patches and classifying each patch by finding its nearest neighbours in the training set. While simple to implement, this approach serves as a useful baseline against which more sophisticated methods can be evaluated.

* **Strengths:**

  Extremely simple to understand and implement with no training phase required. No assumptions about data distribution. Naturally handles multi-class problems. Effective as a baseline method for comparison. Easy to update: adding new training examples requires no retraining.

* **Weaknesses:**

  Very slow at inference time — must compare every test sample to the entire training set. Highly sensitive to the choice of K and the distance metric. Performance degrades sharply in high-dimensional feature spaces (curse of dimensionality). Not suitable for real-time detection on full images. Poor scalability as memory and computation grow linearly with dataset size. Significantly lower accuracy than CNN-based methods on complex, real-world images.

### **Input Data and Pre-processing**

**Data source:**

Sample traffic and vehicle images are downloaded from the COCO dataset validation set and supplemented with publicly available traffic camera images. The COCO dataset (Lin et al., 2014) contains over 330,000 images across 80 object categories, including car, truck, bus, and motorcycle. **Input format:**

Raw RGB images in JPEG or PNG format. Images are variable in size (typically 640×480 to 1920×1080 pixels from camera sources).

**Pre-processing required:**

* Resizing: All images are resized to 640×640 pixels (YOLOv5 default input size) using letterbox padding to preserve aspect ratio without distortion.

* Normalisation: Pixel values are scaled from [0, 255] to [0.0, 1.0] by dividing by 255.

* Channel ordering: Images are loaded in BGR format by OpenCV and converted to RGB for the model.

* Batch dimension: A batch dimension is added (shape becomes [1, 3, 640, 640]) for model inference.

* Confidence filtering: After inference, detections with confidence below 0.25 are discarded.

* Class filtering: Only COCO vehicle class IDs are retained: car (2), motorcycle (3), bus (5), truck (7).

All pre-processing is handled automatically by the Ultralytics YOLOv5 library when images are passed as file paths or NumPy arrays.

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image
import requests
from io import BytesIO
import os
import warnings
warnings.filterwarnings('ignore')

from ultralytics import YOLO

print(f'PyTorch version: {torch.__version__}')
print(f'CUDA available: {torch.cuda.is_available()}')
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using device: {device}')
```

```python
# Load YOLOv5 small model pre-trained on COCO dataset
# 'yolov5s' = small variant, good balance of speed and accuracy for a prototype
model = YOLO('yolov5s.pt')

# COCO vehicle class IDs we care about
VEHICLE_CLASSES = {
    2: 'car',
    3: 'motorcycle',
    5: 'bus',
    7: 'truck'
}

# Colour map for visualisation
CLASS_COLOURS = {
    'car':        '#4A90D9',   # blue
    'motorcycle': '#E8A838',   # amber
    'bus':        '#5BA85A',   # green
    'truck':      '#C0504D'    # red
}

print('YOLOv5s model loaded successfully.')
print(f'Model trained on COCO dataset ({model.names.__len__()} classes)')
print(f'Vehicle classes in scope: {list(VEHICLE_CLASSES.values())}')
```

```python
import subprocess, os, glob, random
from PIL import Image

# Download COCO128 dataset - contains real traffic scenes with ground truth labels
if not os.path.exists('coco128'):
    print('Downloading COCO128 dataset...')
    subprocess.run(['wget', '-q', 'https://ultralytics.com/assets/coco128.zip'])
    subprocess.run(['unzip', '-q', 'coco128.zip'])
    print('COCO128 downloaded.')
else:
    print('COCO128 already present.')

all_coco_images = sorted(glob.glob('coco128/images/train2017/*.jpg'))
print(f'Total images in COCO128: {len(all_coco_images)}')

# Filter to only images confirmed to contain vehicle ground truth labels
# COCO class IDs: car=2, motorcycle=3, bus=5, truck=7
VEHICLE_CLASS_IDS = {'2', '3', '5', '7'}

def has_vehicle_label(img_path):
    """Returns True if this image has at least one vehicle in its label file."""
    stem       = os.path.splitext(os.path.basename(img_path))[0]
    label_path = f'coco128/labels/train2017/{stem}.txt'
    if not os.path.exists(label_path):
        return False
    with open(label_path) as f:
        return any(line.split()[0] in VEHICLE_CLASS_IDS
                   for line in f if line.strip())

vehicle_images = [p for p in all_coco_images if has_vehicle_label(p)]
print(f'Images with confirmed vehicle labels: {len(vehicle_images)}')

# Split into 2 independent sets of 10
# Set 1 is used for Part C detection demo
# Set 2 is used for Part D evaluation on unseen images
random.seed(99)
random.shuffle(vehicle_images)

image_set_1 = vehicle_images[:10]
image_set_2 = vehicle_images[10:20]

print(f'\nImage Set 1 - Part C detection demo ({len(image_set_1)} images):')
for p in image_set_1:
    print(f'  {os.path.basename(p)}')

print(f'\nImage Set 2 - Part D evaluation ({len(image_set_2)} images):')
for p in image_set_2:
    print(f'  {os.path.basename(p)}')

# Set image_paths to Set 1 for the detection cells below
image_paths = image_set_1
print(f'\nimage_paths ready: {len(image_paths)} images from Set 1')
```

```python
def detect_vehicles(model, image_path, conf_threshold=0.25):
    """
    Run YOLOv5 inference on a single image.
    Returns filtered detections containing only vehicle classes.

    Args:
        model: loaded YOLO model
        image_path: path to the input image
        conf_threshold: minimum confidence score to accept a detection

    Returns:
        results: raw YOLO results object
        vehicle_detections: list of dicts with bbox, class, confidence
    """
    results = model(image_path, conf=conf_threshold, verbose=False)
    vehicle_detections = []

    for result in results:
        boxes = result.boxes
        for box in boxes:
            class_id = int(box.cls[0])
            if class_id in VEHICLE_CLASSES:
                x1, y1, x2, y2 = box.xyxy[0].tolist()
                confidence = float(box.conf[0])
                class_name = VEHICLE_CLASSES[class_id]
                vehicle_detections.append({
                    'class_id':   class_id,
                    'class_name': class_name,
                    'confidence': confidence,
                    'bbox':       [x1, y1, x2, y2]
                })

    return results, vehicle_detections


# Run inference on all downloaded images
all_results = []
for path in image_paths:
    results, detections = detect_vehicles(model, path)
    all_results.append({'path': path, 'results': results, 'detections': detections})
    print(f'{path}: {len(detections)} vehicle(s) detected')
    for d in detections:
        print(f'   -> {d["class_name"]:12s} | confidence: {d["confidence"]:.3f}')

print(f'\nInference complete on {len(all_results)} images.')
```

```python
def visualise_detections(image_path, detections, ax):
    """
    Draw bounding boxes and labels on an image using matplotlib.

    Args:
        image_path: path to the original image
        detections: list of detection dicts from detect_vehicles()
        ax: matplotlib Axes object to draw on
    """
    img = Image.open(image_path).convert('RGB')
    ax.imshow(img)
    ax.axis('off')

    for det in detections:
        x1, y1, x2, y2 = det['bbox']
        w = x2 - x1
        h = y2 - y1
        color = CLASS_COLOURS.get(det['class_name'], '#FFFFFF')

        # Draw bounding box rectangle
        rect = patches.Rectangle(
            (x1, y1), w, h,
            linewidth=2.5, edgecolor=color,
            facecolor='none'
        )
        ax.add_patch(rect)

        # Draw label background
        label = f"{det['class_name']} {det['confidence']:.2f}"
        ax.text(
            x1, y1 - 5, label,
            fontsize=8, color='white', fontweight='bold',
            bbox=dict(boxstyle='round,pad=0.2', facecolor=color, alpha=0.85, edgecolor='none')
        )

    title = f'{os.path.basename(image_path)} | {len(detections)} vehicle(s)'
    ax.set_title(title, fontsize=9, pad=4)


# Plot all images with their detections
n_imgs = len(all_results)
cols = 2
rows = (n_imgs + 1) // cols

fig, axes = plt.subplots(rows, cols, figsize=(14, rows * 5))
axes = axes.flatten() if n_imgs > 1 else [axes]

for i, item in enumerate(all_results):
    visualise_detections(item['path'], item['detections'], axes[i])

# Hide unused axes
for j in range(i + 1, len(axes)):
    axes[j].axis('off')

fig.suptitle('YOLOv5 Vehicle Detection Results', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig('detection_results.png', dpi=150, bbox_inches='tight')
plt.show()
print('Detection results plot saved.')
```

```python
from collections import Counter

# Aggregate statistics across all images
all_detections = [d for item in all_results for d in item['detections']]

class_counts = Counter(d['class_name'] for d in all_detections)
confidences = [d['confidence'] for d in all_detections]

print('=== Detection Summary ===')
print(f'Total images processed   : {len(all_results)}')
print(f'Total vehicle detections : {len(all_detections)}')
print()
print('Detections by class:')
for cls, count in sorted(class_counts.items(), key=lambda x: -x[1]):
    print(f'  {cls:15s}: {count}')
print()
if confidences:
    print(f'Mean confidence  : {np.mean(confidences):.4f}')
    print(f'Min confidence   : {np.min(confidences):.4f}')
    print(f'Max confidence   : {np.max(confidences):.4f}')

# --- Bar chart: detections by class ---
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

if class_counts:
    classes = list(class_counts.keys())
    counts  = list(class_counts.values())
    colors  = [CLASS_COLOURS.get(c, '#888888') for c in classes]

    axes[0].bar(classes, counts, color=colors, edgecolor='white', linewidth=1.2)
    axes[0].set_title('Detections by Vehicle Class', fontweight='bold')
    axes[0].set_ylabel('Count')
    axes[0].set_xlabel('Vehicle Class')
    for i, (cls, cnt) in enumerate(zip(classes, counts)):
        axes[0].text(i, cnt + 0.05, str(cnt), ha='center', va='bottom', fontweight='bold')

# --- Confidence score histogram ---
if confidences:
    axes[1].hist(confidences, bins=15, color='#4A90D9', edgecolor='white', linewidth=0.8)
    axes[1].axvline(np.mean(confidences), color='#C0504D', linewidth=2,
                    linestyle='--', label=f'Mean: {np.mean(confidences):.2f}')
    axes[1].set_title('Distribution of Confidence Scores', fontweight='bold')
    axes[1].set_xlabel('Confidence Score')
    axes[1].set_ylabel('Frequency')
    axes[1].legend()

plt.tight_layout()
plt.savefig('detection_statistics.png', dpi=150, bbox_inches='tight')
plt.show()
print('Statistics plot saved.')
```

## Part D - Software Testing & Evaluation

### Testing Method

For object detection systems, the standard testing methodology uses te following metrics:

* Precision measures the proportion of detections that are correct. A high recall means the model in not generating many spurios detections.
* Recall measures the proportion of actual vehiclesthat are deteched. A high recall means the model is not missing many vehicles.
* Intersection over Union (IoU) measures the overlap between a predicted bounding box and the ground truth box. A detection is typically considered correct if IoU >= 0.5.
* Mean Average Precision (mAP) is the primary benchmark metric for object detection. It computes the are under the precision-Recall curve for each class and averages across classes. mAP@0.5 (IoU threshold of 0.5) and mAP@0.5:0.95 (average over thresholds from 0.5 to 0.95) are standard measures.

For this prototype, since a manually labelled ground truth dataset was not available, we evaluate the model using:
1. the published benchmark performance of YOLOv5s on the COCO val2017 dataset,
2. qualitative visual inspection of results on our traffic images,

which constitutes a form of functional testing appropriate for a prototype system.

```python
# ── Part D: Actual Model Evaluation on COCO128 ──────────────────────────────
# Run the model's built-in validation on COCO128 to get REAL metrics
# No hardcoded numbers - everything is measured live

from ultralytics import YOLO
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ── Step 1: Download COCO128 dataset (if not already done) ──────────────────
import subprocess, os

if not os.path.exists('coco128'):
    subprocess.run(['wget', '-q', 'https://ultralytics.com/assets/coco128.zip'])
    subprocess.run(['unzip', '-q', 'coco128.zip'])
    print('COCO128 downloaded.')
else:
    print('COCO128 already present.')

# ── Step 2: Create a minimal dataset YAML so YOLO knows where to look ───────
yaml_content = """
path: coco128
train: images/train2017
val:   images/train2017

nc: 80
names: ['person','bicycle','car','motorcycle','airplane','bus','train','truck',
        'boat','traffic light','fire hydrant','stop sign','parking meter','bench',
        'bird','cat','dog','horse','sheep','cow','elephant','bear','zebra','giraffe',
        'backpack','umbrella','handbag','tie','suitcase','frisbee','skis','snowboard',
        'sports ball','kite','baseball bat','baseball glove','skateboard','surfboard',
        'tennis racket','bottle','wine glass','cup','fork','knife','spoon','bowl',
        'banana','apple','sandwich','orange','broccoli','carrot','hot dog','pizza',
        'donut','cake','chair','couch','potted plant','bed','dining table','toilet',
        'tv','laptop','mouse','remote','keyboard','cell phone','microwave','oven',
        'toaster','sink','refrigerator','book','clock','vase','scissors','teddy bear',
        'hair drier','toothbrush']
"""
with open('coco128.yaml', 'w') as f:
    f.write(yaml_content)
print('Dataset YAML written.')

# ── Step 3: Run actual validation ────────────────────────────────────────────
print('\nRunning validation on COCO128 — this takes ~1 minute on GPU...')
model = YOLO('yolov5s.pt')

metrics = model.val(
    data='coco128.yaml',
    imgsz=640,
    conf=0.25,
    iou=0.5,
    verbose=False,
    plots=False      # we'll make our own plots below
)

# ── Step 4: Extract REAL per-class results ───────────────────────────────────
# Vehicle class IDs in COCO: car=2, motorcycle=3, bus=5, truck=7
VEHICLE_CLASS_IDS   = [2, 3, 5, 7]
VEHICLE_CLASS_NAMES = ['car', 'motorcycle', 'bus', 'truck']

# metrics.box.maps  → mAP@0.5 per class (80 values, one per COCO class)
# metrics.box.p     → precision per class
# metrics.box.r     → recall per class

all_maps  = metrics.box.maps          # shape: (80,)
all_prec  = metrics.box.p             # shape: (80,)  or scalar
all_rec   = metrics.box.r             # shape: (80,)  or scalar

def safe_get(arr, idx):
    """Safely extract a value from array or return scalar."""
    try:
        return float(arr[idx])
    except (TypeError, IndexError):
        return float(arr)

rows = []
for cid, cname in zip(VEHICLE_CLASS_IDS, VEHICLE_CLASS_NAMES):
    rows.append({
        'Class':      cname,
        'Precision':  round(safe_get(all_prec, cid), 4),
        'Recall':     round(safe_get(all_rec,  cid), 4),
        'mAP@0.5':    round(safe_get(all_maps, cid), 4),
        'F1':         round(2 * safe_get(all_prec, cid) * safe_get(all_rec, cid)
                            / max(safe_get(all_prec, cid) + safe_get(all_rec, cid), 1e-9), 4)
    })

df = pd.DataFrame(rows)

# Overall vehicle metrics (mean across the 4 classes)
mean_row = {
    'Class':     'MEAN (vehicles)',
    'Precision':  round(df['Precision'].mean(), 4),
    'Recall':     round(df['Recall'].mean(),    4),
    'mAP@0.5':    round(df['mAP@0.5'].mean(),   4),
    'F1':         round(df['F1'].mean(),         4),
}
df_display = pd.concat([df, pd.DataFrame([mean_row])], ignore_index=True)

print('\n=== YOLOv5s — Actual Performance on COCO128 (vehicle classes) ===\n')
print(df_display.to_string(index=False))
print(f'\nOverall mAP@0.5 (all 80 classes): {metrics.box.map50:.4f}')
print(f'Overall mAP@0.5:0.95            : {metrics.box.map:.4f}')

# ── Step 5: Visualisations ───────────────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

x     = np.arange(len(VEHICLE_CLASS_NAMES))
width = 0.25
colors = ['#4A90D9', '#5BA85A', '#E8A838']

# Chart 1: Precision / Recall / F1 grouped bar per class
b1 = axes[0].bar(x - width, df['Precision'], width, label='Precision', color=colors[0], edgecolor='white')
b2 = axes[0].bar(x,         df['Recall'],    width, label='Recall',    color=colors[1], edgecolor='white')
b3 = axes[0].bar(x + width, df['F1'],        width, label='F1',        color=colors[2], edgecolor='white')
axes[0].set_xticks(x)
axes[0].set_xticklabels(VEHICLE_CLASS_NAMES)
axes[0].set_ylim(0, 1.15)
axes[0].set_title('Precision / Recall / F1 per Class\n(measured on COCO128)', fontweight='bold')
axes[0].set_ylabel('Score')
axes[0].legend()
for bars in [b1, b2, b3]:
    for bar in bars:
        h = bar.get_height()
        axes[0].text(bar.get_x() + bar.get_width()/2, h + 0.02,
                     f'{h:.2f}', ha='center', fontsize=7, fontweight='bold')

# Chart 2: mAP@0.5 per vehicle class
bar_colors = ['#4A90D9', '#E8A838', '#5BA85A', '#C0504D']
bars2 = axes[1].bar(VEHICLE_CLASS_NAMES, df['mAP@0.5'],
                    color=bar_colors, edgecolor='white')
axes[1].set_ylim(0, 1.0)
axes[1].set_title('mAP@0.5 per Vehicle Class\n(measured on COCO128)', fontweight='bold')
axes[1].set_ylabel('mAP@0.5')
axes[1].axhline(df['mAP@0.5'].mean(), color='black', linewidth=1.5,
                linestyle='--', label=f"Mean: {df['mAP@0.5'].mean():.3f}")
axes[1].legend()
for bar, val in zip(bars2, df['mAP@0.5']):
    axes[1].text(bar.get_x() + bar.get_width()/2, val + 0.02,
                 f'{val:.3f}', ha='center', fontweight='bold', fontsize=9)

# Chart 3: Precision-Recall scatter per class
for i, (cname, color) in enumerate(zip(VEHICLE_CLASS_NAMES, bar_colors)):
    axes[2].scatter(df.loc[i,'Recall'], df.loc[i,'Precision'],
                    s=180, color=color, edgecolors='white',
                    linewidth=1.2, zorder=5, label=cname)
    axes[2].annotate(cname,
                     (df.loc[i,'Recall'], df.loc[i,'Precision']),
                     textcoords='offset points', xytext=(8, 4), fontsize=9)
axes[2].set_xlim(-0.05, 1.05)
axes[2].set_ylim(-0.05, 1.05)
axes[2].set_xlabel('Recall')
axes[2].set_ylabel('Precision')
axes[2].set_title('Precision vs Recall per Class\n(measured on COCO128)', fontweight='bold')
axes[2].axhline(0.5, color='gray', linewidth=0.8, linestyle='--', alpha=0.5)
axes[2].axvline(0.5, color='gray', linewidth=0.8, linestyle='--', alpha=0.5)
axes[2].legend(loc='lower left')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('actual_evaluation_metrics.png', dpi=150, bbox_inches='tight')
plt.show()
print('\nEvaluation complete. All metrics are measured, not hardcoded.')
```

## Part E - Equity, Diversity, Inclusion and Sustainability

### Bias, Fairness, Accessibility, and Ethical Implications

The YOLOv5 model was trained on the COCO dataset, which, while large and diverse, was primarily collected from images sourced from Flickr and web scraping, with a bias towards Western urban environments (Lin et al., 2014). This introduces several fairness concerns for a global transportation system:

1. Vehicle type bias:

    COCO includes mainstream vehicle categories (car, truck, bus, motorcycle) but does not include common vehicle types in the Global South such as auto-rickshaws (tuk-tuks), cycle rickshaws, or minibuses. A system deployed in Sri Lanka, India, or sub-Saharan Africa without fine-tuning would likely produce high false-negative rates for locally common vehicle types, potentially causing those vehicle classes to be undercounted in traffic analysis. This could lead to biased infrastructure decisions that disadvantage communities using these vehicles.

2. Environmental bias:

    Images in COCO predominantly reflect daylight conditions in Western cities. The model’s performance in tropical environments (high glare, monsoon rain, red-tinted dust) or at night has not been validated. Systems deployed in these contexts without local evaluation may be unreliable.

3. Surveillance and civil liberties:

    Vehicle detection technology, when combined with licence plate recognition, can enable mass surveillance of citizens’ movements. This raises serious concerns about civil liberties, particularly in contexts without strong data protection legislation. Responsible deployment requires strict access controls, data minimisation, and transparent governance.

### Accessibility and Inclusion

The system outputs are primarily visual (bounding box overlays and matplotlib charts). For visually impaired users or stakeholders, the output should be supplemented with text-based summaries of detections (counts by class, confidence statistics). The current implementation does produce printed text summaries, which partially addresses this. Future versions should include structured JSON output for downstream integration.

### Ethical Implications

Automated vehicle detection can be used for both beneficial and harmful purposes. Beneficial uses include improving traffic flow, reducing accidents, and enabling smarter urban planning. Harmful uses include automated fine enforcement without human review, discriminatory policing that targets certain vehicle types associated with particular communities, and warrantless mass surveillance. Ethical deployment of this technology requires: human oversight of automated decisions, transparency about system limitations, robust appeals mechanisms for false detections, and compliance with national data protection law.

### Sustainability

* Computational resource analysis:

  Training a YOLOv5s model from scratch on COCO (118,000 images, 80 classes, 300 epochs) requires approximately 1.6 GPU-hours on an NVIDIA V100 (Jocher et al., 2022). This corresponds to roughly 0.8 kWh of electricity consumption. At the UK’s grid carbon intensity of approximately 233 gCO2/kWh (National Grid ESO, 2023), this equates to approximately 186g of CO2e per training run — comparable to boiling a kettle 40 times.

  This prototype does not train from scratch but instead uses pre-trained weights, consuming only inference computation. Inference with YOLOv5s on a single image takes 1.98ms on a GPU. Processing 24 hours of traffic camera footage at 25fps would require approximately 21.6 million inferences per day, taking roughly 11.9 hours of GPU time. On a shared cloud GPU, this equates to approximately 5.9 kWh per camera per day, or around 1.37 kg CO2e.

* Data storage requiremnts:

  The YOLOv5s model weights file is 14.1 MB. COCO dataset (used for pre-training) is approximately 25 GB. For deployment, only the model weights are needed. Traffic camera footage storage depends on resolution and compression: at 640x480 H.264, one day of footage per camera is approximately 5–10 GB.

#### Suggestions for Reducing Environemental Impact

1. Model quantisation: Converting the model from float32 to int8 precision reduces memory by 4x and inference energy by approximately 2–3x with minimal accuracy loss. TensorRT or PyTorch Quantization Toolkit can be used for this.

2. Edge deployment: Running inference on edge devices (e.g., NVIDIA Jetson Nano) co-located with cameras eliminates the need to stream high-resolution video to a cloud server, dramatically reducing network bandwidth and associated energy consumption.

3. Event-triggered inference: Rather than processing every frame, the system could use a lightweight motion detector to trigger YOLOv5 inference only when movement is detected in the frame, reducing unnecessary computation by 70–90% in low-traffic periods.

4. Smaller model variants: YOLOv5n (nano) has 1.9M parameters vs. 7.2M for YOLOv5s, with approximately 3.7x lower latency and proportionally lower energy use, at a modest accuracy cost (mAP@0.5: 28.0% vs. 37.4%).

5. Green cloud selection: Deploying on cloud regions powered by renewable energy (e.g., Google Cloud’s Finland or Iowa regions) can reduce carbon emissions by up to 90% compared to coal-heavy grid regions.
